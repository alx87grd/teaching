# Trigo functions

Here are some basic useful function for trigonometry operations

```
# Trigonometry
ang_deg = 30.0
ang_rad = np.radians(ang_deg)
and_deg = np.degrees(ang_rad)
print('Deg: ',ang_deg)
print('Radian: ', ang_rad)

y1 = np.cos(ang_rad)
y3 = np.sin(ang_rad)
y5 = np.tan(ang_rad)

print(y1,y3,y5)
```

```
# Inverse Trigonometric Functions
x1 = np.arccos(y1)
x3 = np.arcsin(y3)
x5 = np.arctan(y5)

print(x1,x3,x5)
```

```
# Atan vs Atan2
ang_deg = -150

s = np.sin(np.radians(ang_deg))
c = np.cos(np.radians(ang_deg))
t = s/c

z1 = np.degrees(np.arctan(t))
z2 = np.degrees(np.arctan2(s, c))

print(f'z1 = {z1}')
print(f'z2 = {z2}')
```

Make sure that you understand what is the difference between arctan and arctan2!!

Arrays in numpy are multi-dimension list of numbers. They can be

### Combining array

```
data = [[1, 2], [3, 4]]
x = np.array(data)

print(f'x =\n {x} \n with shape {x.shape}')
```

Along existing dimensions

```
# Existing dimensions (cat)
print(f'Concatenate along axis 0')
data = np.concatenate([x, x, x], axis=0)
print(f'[x; x; x] =\n {data} \n with shape {data.shape}')

print(f'Concatenate along axis 1')
data = np.concatenate([x, x, x], axis=1)
print(f'[x x x] =\n {data} \n with shape {data.shape}')
```

Stacking along a new dimension

```
# New dimension (stack)
data = np.stack([x, x, x], axis=0)
print(f'\n New array = \n{data} \n with shape {data.shape}')
data = np.stack([x, x, x], axis=1)
print(f'\n New array = \n{data} \n with shape {data.shape}')
data = np.stack([x, x, x], axis=2)
print(f'\n New array = \n{data} \n with shape {data.shape}')
```

```
# Append
u = np.array([[7, 7]])
X1 = np.append(x, u.T, axis=1)  # add column
print(f'With an added column: \n{X1}')
X2 = np.append(x, u, axis=0)  # add row
print(f'With an added row: \n{X2}')
```

```
# Could have been done with concatenate also:
X3 = np.concatenate([x, u.T], axis=1)
print(f'With an added column: \n{X3}')

X4 = np.concatenate([x, u], axis=0)
print(f'With an added row: \n{X4}')
```

## Linear algebra

To manipulate 'row vector', 'column vector' and matrices, like when conducting linear algebra operations, we need to work with 2d-array. Everything is thus a 'matrix': column-vectors are matrices with shape (n,1) and row-vectors are matrices with shape (1,m).


```
data = [[1, 2, 3]] # note the double bracket to create a 2d array
x = np.array(data).T # Column-vector
A = np.eye(3) * 2
A[1,2] = 1

print(f'Matrix A = \n{A}')
print(f'Column vector x = \n{x}')

print(f'Number of dimension of A = {A.ndim}')
print(f'Number of dimension of x = {x.ndim}')

print(f'Shape of A = {A.shape}')
print(f'Shape of x = {x.shape}')
```

Important!

By default, the * operator is element-wise operation multiplication. For matrix multiplication use the operator @

```
# Matrix multipication
y = A @ x
print(f'y = A @ x = \n{y}')
```

Trying to matrix multiply wrong dimension will create an error

```
# error = A @ x.T
```

```
# Scalar product (inner)
xTx = x.T @ x
print(xTx)
```

Note here that using using the * operator here would not lead to the expected result.

```
print(x.T*x)
```

The norm function can also be used to compute $\sum_i x_i^2$


```
print(np.linalg.norm(x) ** 2)
```

```
# Outer/Dyadic product
xxT = x @ x.T
print(xxT)
```

```
# Matrix inversion
A_inv = np.linalg.inv(A)
print( A_inv )

print( A @ A_inv )
```

Pseudo-inverse
$A^{\#}$

```
A_pinv = np.linalg.pinv(A)
print(A_pinv)
```

```
# Solve linear system
y = np.array([[1, 2, 3]]).T
x = A_inv @ y
print(x)
y = A @ x
print(y)
```

```
# Using solve for a linear system of equation
x = np.linalg.solve(A, y)
print(x)
```

```
# Least-square solution
x = np.linalg.lstsq(A, y)[0]
print(x)
```

### Sum/Min/Max

```
data = [[1, 2, 10], [3, 4, 0], [3, 1, 1]]
x = np.array(data)

print(x)
print("Summation per colum: ", np.sum(x, axis=0))
print("Summation per row: ", np.sum(x, axis=1))

print("Min per colum: ", x.min(axis=0))
print("Min per row: ", x.min(axis=1))

print("ArgMin per colum: ", x.argmin(axis=0))
print("ArgMin per row: ", x.argmin(axis=1))

print("Max per colum: ", x.max(axis=0))
print("Max per row: ", x.max(axis=1))

print("ArgMax per colum: ", x.argmax(axis=0))
print("ArgMax per row: ", x.argmax(axis=1))

print("Row index that maximize the sum over column: ", x.sum(axis=0).argmax())
```

### Reshaping

```
tensor = np.array([[[1, 2, 3], [4, 5, 6]],[[2, 2, 2], [3, 3, 3]]])
print(f"Tensor (3D-array):\n {tensor} with shape {tensor.shape}\n")
x = tensor.flatten()
print(f"Vector (1D-array) with the same information:\n {x} with shape {x.shape}\n")
matrix = tensor.reshape(2, -1)  # -1 means automatic
print(f"Matrix (2D-array) with the same information:\n {matrix} with shape {matrix.shape}\n")
tensor = matrix.reshape(2, 2, -1)  # -1 means automatic
print(f"Reconstructed tensor:\n {tensor} with shape {tensor.shape}\n")
```

### Broadcasting for parallel operations

One one avoid for loops with numpy is called "broadcasting", to compute all permutation of (i,j) operation in one line:

```
x = np.array([1,2,3])
y = np.array([0.1,0.2,0.1,0.4,0.5])

z = x[:, np.newaxis] * y[np.newaxis, :]

print(z) # z_ij = x_i * z_j for all (i,j)
```
