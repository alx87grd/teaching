# Plotting

With matplotlib

```
import numpy as np
import matplotlib.pyplot as plt
```

```
import matplotlib.pyplot as plt
ax = plt.figure().add_subplot()

# Plot a sin curve using the x and y axes.
n = 500 # number of point (discretization)
x = np.linspace(0, 1, n)
y = np.sin(x * 2 * np.pi) / 2 + 0.5
ax.plot(x, y)
```

## Basic 3D plots

```
ax = plt.figure().add_subplot(projection='3d')

x = np.linspace(0, 1, 100)
y = np.sin(x * 2 * np.pi) / 2 + 0.5
z = np.sin(x * 4 * np.pi) * 2 + 2.0
ax.plot(x, y, z)
```

### Plotting options



```
n = 100
x = np.linspace(0, 2 * np.pi, n)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.sin(x) * np.cos(x)

fig, ax = plt.subplots(figsize=(8, 6))

# Plotting with different linestyles and colors, and labels
ax.plot(x, y1, label='Sine Wave', color='blue', linestyle='-')
ax.plot(x, y2, label='Cosine Wave', color='red', linestyle='--')
ax.plot(x, y3, label='Sine x Cosine', color='green', linestyle=':')

# Adding labels and title
ax.set_xlabel('X-axis (radians)', fontsize=12)
ax.set_ylabel('Y-axis', fontsize=12)
ax.set_title('Demonstration of Plotting Options', fontsize=14)

# Adding a legend
ax.legend(loc='upper right', fontsize=10)

# Adding a grid
ax.grid(True, linestyle='-', alpha=0.6)

# Customize tick parameters
ax.tick_params(axis='both', which='major', labelsize=10)

plt.tight_layout()
plt.show()
```
